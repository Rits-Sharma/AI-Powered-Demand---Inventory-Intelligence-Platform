```python
import pandas as pd
```

```python
from pathlib import Path
ROOT = Path.cwd().parent
```

```python
sales = pd.read_csv(ROOT / "data" / "raw" / "sales_daily.csv")
sku = pd.read_csv(ROOT / "data" / "raw" / "sku_master.csv")
calendar = pd.read_csv(ROOT / "data" / "raw" / "calendar.csv")
inventory = pd.read_csv(ROOT / "data" / "raw" / "inventory_snapshots.csv")
```

```python
sales.info()
```

```python
datasets = {
    "Sales": sales,
    "SKU Master": sku,
    "Calendar": calendar,
    "Inventory": inventory
}

# Checked missing values, shape, Describe
for name, df in datasets.items():
    print(f"--- Missing Values in {name} ---")
    print("\nNUll values: ", df.isnull().sum())
    print("\nShape:",df.shape)
    print("\nDescribe: ", df.describe())
    print('=' * 80)
```

```python
for name, df  in datasets.items():
    print(name)
    print(df.head(5))
    print("=" * 80)
```

Removed leading spaces

```python
sales.columns = sales.columns.str.strip()
sku.columns = sku.columns.str.strip()
calendar.columns = calendar.columns.str.strip()
inventory.columns = inventory.columns.str.strip()

sales.to_csv(ROOT / "data" / "raw" / "sales_daily.csv", index=False)
sku.to_csv(ROOT / "data" / "raw" / "sku_master.csv", index=False)
calendar.to_csv(ROOT / "data" / "raw" / "calendar.csv", index=False)
inventory.to_csv(ROOT / "data" / "raw" / "inventory_snapshots.csv", index=False)
```

```python
for df in [sales, sku, calendar, inventory]:
    for col in df.select_dtypes(include='object').columns:
        df[col] = df[col].str.strip()
```
